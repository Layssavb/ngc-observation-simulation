```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# ASTRONOMICAL OBSERVATION SIMULATION
# ==========================================================
#
# This script simulates the observational conditions for
# selected NGC galaxies using astronomical relations
# commonly applied in observational astronomy.
#
# The program computes:
#
#   • Object altitude above the horizon
#   • Air mass
#   • Hour angle
#   • Visibility evolution during the night
#
# In addition, the script generates observational graphs:
#
#   • Altitude vs Local Time
#   • Altitude vs Local Sidereal Time
#   • Altitude vs Universal Time
#   • Air Mass vs Time
#
# These quantities are fundamental in observation planning,
# since they directly affect atmospheric extinction,
# image quality, and observational efficiency.
#
# ==========================================================



# ==========================================================
# OBJECT PARAMETERS
# ==========================================================
#
# Right Ascension (RA):
# Angular coordinate measured along the celestial equator.
#
# Declination (Dec):
# Angular coordinate analogous to terrestrial latitude.
#
# Units:
#   • RA  -> hours
#   • Dec -> degrees
#
# ==========================================================

objects = {

    "NGC 10": {
        "RA": 0 + 9/60 + 56.4/3600,
        "Dec": -33 - 42/60 - 34/3600
    },

    "NGC 38": {
        "RA": 0 + 13/60 + 8.1/3600,
        "Dec": -5 - 26/60 - 4/3600
    },

    "NGC 47": {
        "RA": 0 + 15/60 + 52.0/3600,
        "Dec": -7 - 1/60 - 5/3600
    }
}



# ==========================================================
# CONVERSION TO RADIANS
# ==========================================================
#
# Trigonometric functions in NumPy use radians.
#
# Conversion factors:
#
#   RA:
#       24h  → 2π rad
#
#   Dec:
#       180° → π rad
#
# ==========================================================

for obj in objects.values():

    obj["RA"] *= np.pi / 12
    obj["Dec"] *= np.pi / 180



# ==========================================================
# OBSERVATORY PARAMETERS
# ==========================================================
#
# Observatory:
# Cerro Tololo Inter-American Observatory
#
# Latitude:
# Important because object altitude depends directly
# on the observer's geographical latitude.
#
# ==========================================================

latitude = -30.165 * np.pi / 180



# ==========================================================
# TIME ARRAYS
# ==========================================================
#
# LT  -> Local Time
# UT  -> Universal Time
# LST -> Local Sidereal Time
#
# The simulation covers the interval from:
#
#       19:00 → 06:00
#
# ==========================================================

LT = np.linspace(19, 30, 200)

# Approximate Chilean local time conversion
UT = LT + 4

# Approximate Local Sidereal Time
LST0 = 0.465
LST = LST0 + (LT - 24) * 1.0027379



# ==========================================================
# FUNCTIONS
# ==========================================================

def altitude(dec, HA):

    """
    Compute object altitude above the horizon.

    Formula:

        sin(h) =
            sin(phi) sin(dec)
            + cos(phi) cos(dec) cos(HA)

    where:

        h   -> altitude
        phi -> observatory latitude
        dec -> declination
        HA  -> hour angle

    """

    return np.arcsin(
        np.sin(latitude) * np.sin(dec)
        + np.cos(latitude) * np.cos(dec) * np.cos(HA)
    )



def air_mass(alt):

    """
    Compute air mass using the secant approximation.

    Air mass quantifies how much atmosphere the light
    crosses before reaching the telescope.

    Important observations:

        • X ≈ 1  -> object near zenith
        • Larger X -> stronger atmospheric effects
        • X > 2  -> usually poor observational quality

    """

    zenith_distance = np.pi / 2 - alt

    return 1 / np.cos(zenith_distance)



# ==========================================================
# MAIN OBSERVATIONAL LOOP
# ==========================================================

for name, obj in objects.items():

    # ======================================================
    # HOUR ANGLE
    # ======================================================
    #
    # HA measures the angular distance between the object
    # and the local meridian.
    #
    # HA = 0:
    #   Object at culmination (maximum altitude)
    #
    # ======================================================

    HA = (LST * np.pi / 12) - obj["RA"]



    # ======================================================
    # ALTITUDE CALCULATION
    # ======================================================

    alt = altitude(obj["Dec"], HA)

    # Convert radians → degrees
    alt_deg = alt * 180 / np.pi



    # ======================================================
    # AIR MASS
    # ======================================================

    X = air_mass(alt)



    # ======================================================
    # REMOVE VALUES BELOW HORIZON
    # ======================================================
    #
    # Negative altitudes correspond to objects below
    # the horizon and therefore not observable.
    #
    # ======================================================

    alt_deg[alt_deg < 0] = np.nan
    X[alt_deg < 0] = np.nan



    # ======================================================
    # ALTITUDE GRAPHS
    # ======================================================

    # ------------------------------------------------------
    # Altitude vs Local Time
    # ------------------------------------------------------

    plt.figure()

    plt.plot(LT, alt_deg, color='red')

    plt.xlabel("Local Time (LT)")
    plt.ylabel("Altitude (degrees)")

    plt.title(f"{name} — Altitude vs Local Time")

    plt.grid()

    plt.show()



    # ------------------------------------------------------
    # Altitude vs Local Sidereal Time
    # ------------------------------------------------------

    plt.figure()

    plt.plot(LST, alt_deg, color='blue')

    plt.xlabel("Local Sidereal Time (LST)")
    plt.ylabel("Altitude (degrees)")

    plt.title(f"{name} — Altitude vs LST")

    plt.grid()

    plt.show()



    # ------------------------------------------------------
    # Altitude vs Universal Time
    # ------------------------------------------------------

    plt.figure()

    plt.plot(UT, alt_deg, color='green')

    plt.xlabel("Universal Time (UT)")
    plt.ylabel("Altitude (degrees)")

    plt.title(f"{name} — Altitude vs UT")

    plt.grid()

    plt.show()



    # ======================================================
    # AIR MASS GRAPHS
    # ======================================================

    # ------------------------------------------------------
    # Air Mass vs Local Time
    # ------------------------------------------------------

    plt.figure()

    plt.plot(LT, X, color='orange')

    plt.xlabel("Local Time (LT)")
    plt.ylabel("Air Mass (X)")

    plt.title(f"{name} — Air Mass vs Local Time")

    plt.grid()

    plt.show()



    # ------------------------------------------------------
    # Air Mass vs Local Sidereal Time
    # ------------------------------------------------------

    plt.figure()

    plt.plot(LST, X, color='purple')

    plt.xlabel("Local Sidereal Time (LST)")
    plt.ylabel("Air Mass (X)")

    plt.title(f"{name} — Air Mass vs LST")

    plt.grid()

    plt.show()



    # ------------------------------------------------------
    # Air Mass vs Universal Time
    # ------------------------------------------------------

    plt.figure()

    plt.plot(UT, X, color='pink')

    plt.xlabel("Universal Time (UT)")
    plt.ylabel("Air Mass (X)")

    plt.title(f"{name} — Air Mass vs UT")

    plt.grid()

    plt.show()



# ==========================================================
# FINAL OBSERVATIONS
# ==========================================================
#
# The generated graphs allow identification of:
#
#   • Best observing intervals
#   • Object culmination times
#   • Lowest air mass conditions
#   • Visibility duration
#
# Observations performed at larger altitudes and
# lower air masses generally provide:
#
#   • Better image quality
#   • Lower atmospheric extinction
#   • Higher observational precision
#
# ==========================================================
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# COMPARATIVE ASTRONOMICAL OBSERVATION SIMULATION
# ==========================================================
#
# This script compares the observational conditions of
# multiple NGC galaxies throughout the night.
#
# The program computes:
#
#   • Object altitude
#   • Air mass
#   • Visibility intervals
#   • Temporal evolution of observational conditions
#
# Comparative plots are generated to identify:
#
#   • Best observing targets
#   • Highest culmination objects
#   • Lowest air mass conditions
#   • Optimal observing windows
#
# ==========================================================



# ==========================================================
# OBJECT PARAMETERS
# ==========================================================
#
# Each object contains:
#
#   • Right Ascension (RA)
#   • Declination (Dec)
#   • Plot color
#
# Units:
#
#   • RA  -> hours
#   • Dec -> degrees
#
# ==========================================================

objects = {

    "NGC 10": {

        "RA": 0 + 9/60 + 56.4/3600,
        "Dec": -33 - 42/60 - 34/3600,

        "color": "red"
    },

    "NGC 38": {

        "RA": 0 + 13/60 + 8.1/3600,
        "Dec": -5 - 26/60 - 4/3600,

        "color": "brown"
    },

    "NGC 47": {

        "RA": 0 + 15/60 + 52.0/3600,
        "Dec": -7 - 1/60 - 5/3600,

        "color": "blue"
    }
}



# ==========================================================
# CONVERSION TO RADIANS
# ==========================================================
#
# Trigonometric calculations require radian units.
#
# Conversion:
#
#   RA:
#       24h  → 2π rad
#
#   Dec:
#       180° → π rad
#
# ==========================================================

for obj in objects.values():

    obj["RA"] *= np.pi / 12
    obj["Dec"] *= np.pi / 180



# ==========================================================
# OBSERVATORY PARAMETERS
# ==========================================================
#
# Observatory:
# Cerro Tololo Inter-American Observatory
#
# Latitude strongly affects:
#
#   • Maximum altitude
#   • Visibility duration
#   • Air mass
#
# ==========================================================

latitude = -30.165 * np.pi / 180



# ==========================================================
# TIME ARRAYS
# ==========================================================
#
# LT  -> Local Time
# UT  -> Universal Time
# LST -> Local Sidereal Time
#
# The simulation interval corresponds to:
#
#       19:00 → 06:00
#
# ==========================================================

LT = np.linspace(19, 30, 300)

# Approximate Chilean local time conversion
UT = LT + 4

# Approximate Local Sidereal Time
LST0 = 0.465
LST = LST0 + (LT - 24) * 1.0027379



# ==========================================================
# FUNCTIONS
# ==========================================================

def altitude(dec, HA):

    """
    Compute object altitude above the horizon.

    Astronomical relation:

        sin(h) =
            sin(phi) sin(dec)
            + cos(phi) cos(dec) cos(HA)

    where:

        h   -> altitude
        phi -> observatory latitude
        dec -> declination
        HA  -> hour angle

    """

    return np.arcsin(
        np.sin(latitude) * np.sin(dec)
        + np.cos(latitude) * np.cos(dec) * np.cos(HA)
    )



def air_mass(alt):

    """
    Compute air mass using the secant approximation.

    Important interpretation:

        • X ≈ 1:
            object near zenith

        • Larger X:
            stronger atmospheric extinction

        • X > 2:
            usually poor observational quality

    """

    zenith_distance = np.pi / 2 - alt

    return 1 / np.cos(zenith_distance)



# ==========================================================
# COMPUTE OBSERVATIONAL PARAMETERS
# ==========================================================

results = {}

for name, obj in objects.items():

    # ------------------------------------------------------
    # Hour Angle
    # ------------------------------------------------------

    HA = (LST * np.pi / 12) - obj["RA"]


    # ------------------------------------------------------
    # Altitude
    # ------------------------------------------------------

    alt = altitude(obj["Dec"], HA)

    alt_deg = alt * 180 / np.pi


    # ------------------------------------------------------
    # Air Mass
    # ------------------------------------------------------

    X = air_mass(alt)


    # ------------------------------------------------------
    # Remove values below the horizon
    # ------------------------------------------------------

    alt_deg[alt_deg < 0] = np.nan
    X[alt_deg < 0] = np.nan


    # ------------------------------------------------------
    # Store results
    # ------------------------------------------------------

    results[name] = {

        "altitude": alt_deg,
        "air_mass": X
    }



# ==========================================================
# 1) ALTITUDE vs LOCAL TIME
# ==========================================================
#
# This graph shows how object altitude changes
# throughout the night in local time coordinates.
#
# Higher altitude:
#
#   • Lower atmospheric extinction
#   • Better image quality
#   • Lower air mass
#
# ==========================================================

plt.figure(figsize=(8, 5))

for name, obj in objects.items():

    plt.plot(
        LT,
        results[name]["altitude"],

        label=name,
        color=obj["color"],
        linewidth=2
    )

plt.xlabel("Local Time (LT)")
plt.ylabel("Altitude (degrees)")

plt.title("Altitude vs Local Time")

plt.legend()

plt.grid()

plt.show()



# ==========================================================
# 2) ALTITUDE vs LOCAL SIDEREAL TIME
# ==========================================================
#
# Sidereal time is directly related to Earth's rotation
# relative to the celestial sphere.
#
# Astronomers frequently use LST to determine:
#
#   • Object culmination
#   • Transit times
#   • Telescope pointing
#
# ==========================================================

plt.figure(figsize=(8, 5))

for name, obj in objects.items():

    plt.plot(
        LST,
        results[name]["altitude"],

        label=name,
        color=obj["color"],
        linewidth=2
    )

plt.xlabel("Local Sidereal Time (LST)")
plt.ylabel("Altitude (degrees)")

plt.title("Altitude vs LST")

plt.legend()

plt.grid()

plt.show()



# ==========================================================
# 3) ALTITUDE vs UNIVERSAL TIME
# ==========================================================
#
# Universal Time (UT) is the international reference
# time standard used in astronomy.
#
# ==========================================================

plt.figure(figsize=(8, 5))

for name, obj in objects.items():

    plt.plot(
        UT,
        results[name]["altitude"],

        label=name,
        color=obj["color"],
        linewidth=2
    )

plt.xlabel("Universal Time (UT)")
plt.ylabel("Altitude (degrees)")

plt.title("Altitude vs UT")

plt.legend()

plt.grid()

plt.show()



# ==========================================================
# 4) AIR MASS vs LOCAL TIME
# ==========================================================
#
# Lower air mass values indicate better atmospheric
# observing conditions.
#
# Typical good observational condition:
#
#       X < 1.55
#
# ==========================================================

plt.figure(figsize=(8, 5))

for name, obj in objects.items():

    plt.plot(
        LT,
        results[name]["air_mass"],

        label=name,
        color=obj["color"],
        linewidth=2
    )

plt.xlabel("Local Time (LT)")
plt.ylabel("Air Mass (X)")

plt.title("Air Mass vs Local Time")

plt.xlim(28.5, 31)

plt.legend()

plt.grid()

plt.show()



# ==========================================================
# 5) AIR MASS vs LOCAL SIDEREAL TIME
# ==========================================================

plt.figure(figsize=(8, 5))

for name, obj in objects.items():

    plt.plot(
        LST,
        results[name]["air_mass"],

        label=name,
        color=obj["color"],
        linewidth=2
    )

plt.xlabel("Local Sidereal Time (LST)")
plt.ylabel("Air Mass (X)")

plt.title("Air Mass vs LST")

plt.xlim(4.5, 7)

plt.legend()

plt.grid()

plt.show()



# ==========================================================
# 6) AIR MASS vs UNIVERSAL TIME
# ==========================================================

plt.figure(figsize=(8, 5))

for name, obj in objects.items():

    plt.plot(
        UT,
        results[name]["air_mass"],

        label=name,
        color=obj["color"],
        linewidth=2
    )

plt.xlabel("Universal Time (UT)")
plt.ylabel("Air Mass (X)")

plt.title("Air Mass vs UT")

plt.xlim(32.5, 35)

plt.legend()

plt.grid()

plt.show()



# ==========================================================
# FINAL COMMENTS
# ==========================================================
#
# The comparative analysis allows identification of:
#
#   • Which object reaches the highest altitude
#   • Which object presents the lowest air mass
#   • Best observing intervals
#   • Relative visibility quality
#
# Objects observed:
#
#   • At larger altitudes
#   • Near culmination
#   • With smaller air mass
#
# generally provide better scientific data quality.
#
# ==========================================================
```
